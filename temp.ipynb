```python
import os
from PIL import Image

def remove_metadata_from_images(directory):
    for root, _, files in os.walk(directory):
        for file in files:
            if file.lower().endswith(('.png', '.jpg', '.jpeg', '.tiff', '.bmp', '.gif', '.PNG', '.JPG', '.JPEG', '.TIFF', '.BMP', '.GIF')):
                file_path = os.path.join(root, file)
                try:
                    with Image.open(file_path) as img:
                        data = list(img.getdata())
                        img_no_meta = Image.new(img.mode, img.size)
                        img_no_meta.putdata(data)
                        img_no_meta.save(file_path)
                        print(f"Removed metadata from {file_path}")
                except Exception as e:
                    print(f"Failed to process {file_path}: {e}")
    print("Metadata removal complete.")

# Example usage:
remove_metadata_from_images('D:\\Datasets Main\\3. occlusal contact segmentation\\(Occlusal-SEG)_Occlusal_Contact_Dataset\\Original_AP_Masks')
```
